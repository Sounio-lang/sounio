# Clinical Ontology Demo

This notebook is the first public proof for Phase 1 of the 15M ontology integration. It resolves mixed biomedical codes through the Sounio ontology subsystem, normalizes a small patient payload, and exposes mapping confidence and provenance.


```python
import json
from pathlib import Path
import sounio
```

## Resolve a clinical diagnosis from SNOMED


```python
%ontology_resolve SNOMED:44054006
```

## Search the mixed ontology slice


```python
%ontology_search glucose
```

## Normalize a patient payload across SNOMED, LOINC, HPO, GO, and ChEBI


```python
fixture = Path("clinical_patient_fixture.json")
payload = json.loads(fixture.read_text())
normalized = sounio.ontology.clinical_normalize(payload)
print(json.dumps(normalized, indent=2))
```

## Compiler-backed validation style checks

These are the exact semantic questions we want the self-hosted compiler CLI and stdlib facade to answer.


```python
resolver = sounio.ontology.get_resolver()
print("ancestors:", resolver.ancestors("SNOMED:44054006"))
print("is_subclass:", resolver.is_subclass("SNOMED:44054006", "SNOMED:73211009"))
print("map_term -> HPO:", [m.to_dict() for m in resolver.map_term("SNOMED:44054006", "HPO")])
print("map_term -> CHEBI:", [m.to_dict() for m in resolver.map_term("SNOMED:44054006", "CHEBI")])
```
